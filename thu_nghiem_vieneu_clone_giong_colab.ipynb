Notebook này chạy trên Google Colab. Trong **Runtime → Change runtime type**, chọn GPU rồi chạy từng ô từ trên xuống. Dùng audio tham chiếu mà bạn có quyền sử dụng; nên chọn đoạn giọng rõ, ít tạp âm, khoảng 3–8 giây. VieNeu nhận audio tham chiếu trực tiếp, không cần nhập transcript.

```python
import torch
from google.colab import files

print("CUDA khả dụng:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Không thấy GPU; VieNeu vẫn có thể chạy CPU nhưng sẽ chậm hơn.")
```

```python
uploaded = files.upload()
if not uploaded:
    raise ValueError("Hãy tải lên một tệp audio tham chiếu.")
reference_audio_path = next(iter(uploaded))
print("Đã chọn audio tham chiếu:", reference_audio_path)
```

```python
from vieneu import Vieneu

tts = Vieneu(mode="v3turbo")
print("Đã nạp VieNeu v3 Turbo.")
```

```python
text_to_speak = "Xin chào, đây là đoạn kiểm tra khả năng clone giọng tiếng Việt bằng VieNeu v3 Turbo."
audio = tts.infer(text_to_speak, ref_audio=reference_audio_path, denoise=True)
output_path = "vieneu_clone_thu.wav"
tts.save(audio, output_path)
print("Đã lưu:", output_path)
```

```python
from IPython.display import Audio, display

display(Audio(filename=output_path))
```
